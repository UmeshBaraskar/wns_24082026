```python
dbutils.widgets.combobox(
    name="top_n_records",
    defaultValue="5",
    choices=["5", "10", "15"],
    label="Top N Records",
)
```

```python
print(dbutils.widgets.get("top_n_records"))
```

```python
dbutils.widgets.dropdown(
    name="technology",
    defaultValue="4G",
    choices=["2G", "4G", "5G", "3G", "VOLTE"],
    label="Technology",
)
```

```python
Top_n_records = dbutils.widgets.get("top_n_records")
print(dbutils.widgets.get("technology"))
```

```python
dbutils.widgets.dropdown(name = "Designation",
                           defaultValue = "Writer",
                           choices = ["Writer", "Artist","Other","Engineer","salesman","Entertainment","Programmer","technician","Student"],
                           label = 'Designation')
```

```python
Designation = dbutils.widgets.get("Designation")
```

```python
from pyspark.sql.functions import col
dfr = spark.read
df = dfr.csv(
    "/Volumes/wns25082026/quick_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    inferSchema = True,
)
df.groupBy("designation").count().sort("count",ascending = False).filter(col("designation")==Designation).display()

```
