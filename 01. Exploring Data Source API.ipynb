### Attaching customer Schema

```python
from pyspark.sql.types import *

user_schema = StructType(
    [
        StructField("id", IntegerType()),
        StructField("age", IntegerType()),
        StructField("gender", StringType()),
        StructField("designation", StringType()),
        StructField("salary", IntegerType()),
    ]
)

spark.read.csv(
    "/Volumes/wns25082026/quick_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    schema=user_schema,
).display()
```

# Handling Bad Records
>>
1. PERMISSIVE (DEFAULT)
2. DROPMALFROMED
3. FAILFAST

```python
spark.read.option("columnNameOfCorruptRecord","Bad_Record").json("/Volumes/wns25082026/quick_schema/sandbox/datasets/access_logs.json", 
                mode = "PERMISSIVE").display()
```

```python
spark.read.json("/Volumes/wns25082026/quick_schema/sandbox/datasets/access_logs.json", 
                mode = "DROPMALFORMED").display()
```

```python
spark.read.json("/Volumes/wns25082026/quick_schema/sandbox/datasets/access_logs.json", 
                mode = "FAILFAST").display()
```

#DataFrame Writer API

## Output modes
1. errorIfExist (default)
2. Overwrite
3. Append
4. Ignore

```python
spark.read.csv(
    "/Volumes/wns25082026/quick_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    inferSchema=True,
).write.save(
    "/Volumes/wns25082026/quick_schema/sandbox/datasets/output/output_parquet",
    mode="OVERWRITE",
    format="PARQUET",
)
```

```python
spark.read.csv(
    "/Volumes/wns25082026/quick_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    inferSchema=True,
).write.save(
    "/Volumes/wns25082026/quick_schema/sandbox/datasets/output/output_parquet",
    mode="APPEND",
    format="PARQUET",
)
```

```python
spark.read.csv(
    "/Volumes/wns25082026/quick_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    inferSchema=True,
).write.save(
    "/Volumes/wns25082026/quick_schema/sandbox/datasets/output/output_parquet",
    mode="IGNORE",
    format="PARQUET",
)
```
